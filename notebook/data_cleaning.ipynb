```python
df['Offer'].value_counts()
```

```python
# Offer Identifies the last marketing offer that the customer accepted: None, Offer A, Offer B, Offer C, Offer D, Offer E
```

```python
df['Offer'] = df['Offer'].fillna('none')
```

```python
df['Offer'].head(10)
```

```python
df['Avg Monthly Long Distance Charges'].sample(10)
```

```python
# # Filled structural missing values with 0 as customers without Phone Service incur no long-distance charges.
df['Avg Monthly Long Distance Charges'] = df['Avg Monthly Long Distance Charges'].fillna(0)
```

```python
df['Multiple Lines'].head()
```

```python
# Verified that missing values occur only when Phone Service = 'No'; filled as per business rule.
df[df['Phone Service'] == 'No'][['Multiple Lines', 'Phone Service']].sample(10)
```

```python
df['Multiple Lines'] = df['Multiple Lines'].fillna('No')
```

```python
# Verified that missing values occur only when Internet Service = 'No'; retained as structural missing values.
df[df['Internet Service'] == 'No'][['Internet Type', 'Internet Service']].sample(10)
```

```python
df['Internet Type'] = df['Internet Type'].fillna('none')
```

```python
# Verified business rule; replaced structural missing values with 0.
df['Avg Monthly GB Download'] = df['Avg Monthly GB Download'].fillna(0)
```

```python
df['Online Security'].unique
```

```python
df[df['Internet Service'] == 'No'][['Online Backup', 'Internet Service']].sample(10)
```

```python
df['Online Security'] = df['Online Security'].fillna('No')
```

```python
# Filled structural missing values with 'No' for customers without Internet Service.
df['Online Backup'] = df['Online Backup'].fillna('No')
df['Device Protection Plan'] = df['Device Protection Plan'].fillna('No')
df['Premium Tech Support'] = df['Premium Tech Support'].fillna('No')
df['Streaming TV'] = df['Streaming TV'].fillna('No')
df['Streaming Movies'] = df['Streaming Movies'].fillna('No')
df['Streaming Music'] = df['Streaming Music'].fillna('No')
df['Unlimited Data'] = df['Unlimited Data'].fillna('No')
```

```python
df.isnull().sum()
```

```python
# Validated business rules using cross-tabulation before handling structural missing values.
internet_cols = [
    'Online Security',
    'Online Backup',
    'Device Protection Plan',
    'Premium Tech Support',
    'Streaming TV',
    'Streaming Movies',
    'Streaming Music',
    'Unlimited Data'
]

for col in internet_cols:
    print(f"\n{'='*50}")
    print(f"{col}")
    print(pd.crosstab(
        df['Internet Service'],
        df[col],
        dropna=False,
        margins=True
    ))
```

```python
df[['Churn Category','Churn Reason','Customer Status']].value_counts()              
```

```python
pd.crosstab(df['Customer Status'], df['Churn Category'], dropna=False, margins=True)
```

```python
df['Churn Category'].value_counts()
```

```python
df['Churn Category'] = df['Churn Category'].fillna('not_churned')
```

```python
pd.crosstab(df['Customer Status'], df['Churn Reason'], dropna=False, margins=True)
```

```python
df['Churn Reason'].head(10)
```

```python
df['Churn Reason'] = df['Churn Reason'].fillna('not_churned')
```

# Data Cleaning Summary

The telecom customer dataset was systematically cleaned and validated to ensure high data quality before exploratory analysis, SQL querying, dashboard development, and predictive modeling.

### Cleaning Activities Performed

- Successfully merged the customer and population datasets using **Zip Code**.
- Investigated all missing values using the **data dictionary** before applying any transformations.
- Identified and handled **structural missing values** based on documented business rules:
  - Filled **Multiple Lines** with `"No"` for customers without Phone Service.
  - Filled **Average Monthly Long Distance Charges** with `0` for customers without Phone Service.
  - Filled internet-dependent service columns (`Online Security`, `Online Backup`, `Device Protection Plan`, `Premium Tech Support`, `Streaming TV`, `Streaming Movies`, `Streaming Music`, and `Unlimited Data`) with `"No"` for customers without Internet Service.
  - Filled **Average Monthly GB Download** with `0` for customers without Internet Service.
- Verified through **cross-tabulation** that all imputations were consistent with the business rules defined in the data dictionary.
- Confirmed that **Churn Category** and **Churn Reason** are recorded only for customers whose **Customer Status** is **Churned**. Since the missing values represent **not applicable** cases rather than missing information, they were retained (or documented appropriately if display columns were created).
- Verified that the dataset contains **no duplicate records** and that each **Customer ID** uniquely represents a single customer.
- Validated and corrected data types where required by converting **Zip Code** from an integer to a string, as it represents a location identifier rather than a numerical measure.
- Performed data quality checks to ensure the dataset is internally consistent and ready for downstream analysis.

### Outcome

The dataset is now clean, consistent, and well-documented, providing a reliable foundation for SQL analysis, exploratory data analysis (EDA), machine learning, and Power BI dashboard development.
